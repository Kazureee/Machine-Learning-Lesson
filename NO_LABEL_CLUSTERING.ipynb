```python
from google.colab import drive
drive.mount('/content/drive')

import pandas as pd
from matplotlib import pyplot as plt
```

```python
df = pd.read_csv('/content/drive/MyDrive/Colab Notebooks/data/Mall_Customers.csv')
df.head(5)
```

```python


plt.scatter(df['Annual Income (k$)'], df['Spending Score (1-100)'])
plt.xlabel('Annual Income (k$)')
plt.ylabel('Spending Score (1-100)')
plt.show()
```

```python
#creating a label

subset = df[['Annual Income (k$)','Spending Score (1-100)']]
subset.head(5)
```

```python
from sklearn.cluster import AgglomerativeClustering
hc = AgglomerativeClustering(n_clusters=5, linkage='ward')
y_predict = hc.fit_predict(subset)


df["Cluster"] = y_predict
df.head()

pd.to_csv(df)
```

```python
df1 = df[df['Cluster'] == 0]
df2 = df[df['Cluster'] == 1]
df3 = df[df['Cluster'] == 2]
df4 = df[df['Cluster'] == 3]
df5 = df[df['Cluster'] == 4]

plt.scatter(df1['Annual Income (k$)'], df1['Spending Score (1-100)'], color='green')
plt.scatter(df2['Annual Income (k$)'], df2['Spending Score (1-100)'], color='blue')
plt.scatter(df3['Annual Income (k$)'], df3['Spending Score (1-100)'], color='red')
plt.scatter(df4['Annual Income (k$)'], df4['Spending Score (1-100)'], color='yellow')
plt.scatter(df5['Annual Income (k$)'], df5['Spending Score (1-100)'], color='black')
plt.xlabel('Annual Income (k$)')
plt.ylabel('Spending Score (1-100)')
plt.show()
```
